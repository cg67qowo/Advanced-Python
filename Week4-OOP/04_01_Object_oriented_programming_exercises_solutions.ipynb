### Exercise 1 — Create a `Student` class

Create a class called `Student` with:

- `name`
- `course`
- a method called `introduce()` that prints a sentence introducing the student.

Then create **two different students** and call `introduce()` for each.

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm studying {self.course}.")

student1 = Student("Alex", "Python")
student2 = Student("Sam", "Data Science")

student1.introduce()
student2.introduce()
```

### Exercise 2 — Add behavior

Extend `Student` with a method `study(hours)` that prints a message such as:

`Alex studied Python for 2 hours.`

**Challenge:** What happens if you call the method with `hours=0`?

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm studying {self.course}.")

    def study(self, hours):
        print(f"{self.name} studied {self.course} for {hours} hours.")

student1 = Student("Alex", "Python")
student1.study(2)
student1.study(0)   # challenge
```

**Answer to the challenge.** Nothing breaks: the method simply prints `Alex studied Python for 0 hours.`
Python does not know that 0 hours is odd; if we want to treat that case (or negative values) specially, we have to test it ourselves,
and we may also want correct grammar for 1 hour. One possible version:

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm studying {self.course}.")

    def study(self, hours):
        if hours <= 0:
            print(f"{self.name} did not study {self.course}.")
        else:
            unit = "hour" if hours == 1 else "hours"
            print(f"{self.name} studied {self.course} for {hours} {unit}.")

student1 = Student("Alex", "Python")
student1.study(2)
student1.study(1)
student1.study(0)
```

### Exercise 3 — Class variable

Create a `Book` class with:

- an instance variable `title`
- an instance variable `author`
- a class variable `category = "Books"`

Create two books and print their titles and the shared category.

**Challenge:** Change the category through `Book.category` and observe what happens.

```python
class Book:
    category = "Books"          # class variable, shared

    def __init__(self, title, author):
        self.title = title      # instance variables
        self.author = author

book1 = Book("Dune", "Frank Herbert")
book2 = Book("Emma", "Jane Austen")

print(book1.title, "-", book1.category)
print(book2.title, "-", book2.category)

# Challenge
Book.category = "Novels"
print(book1.category)   # Novels
print(book2.category)   # Novels
```

**Answer to the challenge.** Both instances now print `Novels`: they do not store the value themselves, Python finds it in the class.

*Bonus:* assigning through an **instance** does something different. It creates a new *instance* variable that hides the class variable for that object only.

```python
book1.category = "Favourites"

print(book1.category)   # Favourites (instance variable, hides the class one)
print(book2.category)   # Novels     (still the class variable)
print(Book.category)    # Novels
```

### Exercise 4 — Build a `Counter`

Create a `Counter` class with:

- an instance variable `value`, starting at `0`
- a method `increment()` that adds 1
- a method `decrement()` that subtracts 1
- a method `show()` that prints the current value

Create two counters and verify that they keep separate values.

```python
class Counter:
    def __init__(self):
        self.value = 0

    def increment(self):
        self.value += 1

    def decrement(self):
        self.value -= 1

    def show(self):
        print(self.value)

c1 = Counter()
c2 = Counter()

c1.increment()
c1.increment()
c1.increment()
c2.decrement()

c1.show()   # 3
c2.show()   # -1
```

Each counter has its own `value` (an instance variable), so changing one never affects the other.
