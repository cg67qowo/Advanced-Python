### 4.1.1 A first look at classes

A class is a **blueprint** for creating objects.

For example, imagine a `Dog` class. Each dog can have:

- **data**: name, age, breed
- **behavior**: bark, sleep, eat

An object created from the class is called an **instance**.

````python
class Dog:
    pass

buddy = Dog()
luna = Dog()

print(type(buddy))
print(type(luna))
````

4.1.2 The `__init__` method

Most beginner classes need an initializer. Python calls `__init__` automatically when an object is created.

The first parameter is conventionally named `self`. It refers to the **current instance**.

When we write:

```python
buddy = Dog("Buddy", 3)
```

Python creates a `Dog` object and calls its `__init__` method with that object as `self`.

````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

buddy = Dog("Buddy", 3)
luna = Dog("Luna", 5)

print(buddy.name)
print(luna.age)
````

### 4.1.2 Class and instance variables


**Instance variables** belong to a particular object.

In this example, each `Dog` has its own `name` and `age`:

```python
buddy.name
luna.name
```

Changing Buddy's name does not change Luna's name.

````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

buddy = Dog("Buddy", 3)
luna = Dog("Luna", 5)

buddy.name = "Max"

print(buddy.name)
print(luna.name)
````

A class variable is shared by the class and, normally, by its instances.

It is useful for information that should be common to all objects.

````python
class Dog:
    species = "Canis familiaris"  # class variable

    def __init__(self, name):
        self.name = name            # instance variable

buddy = Dog("Buddy")
luna = Dog("Luna")

print(buddy.species)
print(luna.species)
print(Dog.species)
````

### 4.1.3 Inheritance


Inheritance allows a new class to reuse behavior from an existing class.

For example:

- `Animal` → general behavior
- `Dog` → a more specific kind of animal
- `Cat` → another specific kind of animal

The child class is often called a **subclass**; the parent class is the **base class** or **superclass**.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} says: Woof!")

dog = Dog("Buddy")

dog.eat()   # inherited
dog.bark()  # defined in Dog
````
