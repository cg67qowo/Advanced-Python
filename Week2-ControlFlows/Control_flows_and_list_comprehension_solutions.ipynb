##### Exercise 3

Replace `???` with `pass`, `continue`, or `break`.

A. Print every number:

````python
for number in range(5):
    if number == 2:
        ???

    print(number)
````

B. Skip `2`:

````python
for number in range(5):
    if number == 2:
        ???

    print(number)
````

C. Stop when reaching `2`:

````python
for number in range(5):
    if number == 2:
        ???

    print(number)
````

````python
# Solution
# Solution A
for number in range(5):
    if number == 2:
        pass
    print(number)

print("---")

# Solution B
for number in range(5):
    if number == 2:
        continue
    print(number)

print("---")

# Solution C
for number in range(5):
    if number == 2:
        break
    print(number)
````

### 2.4 Looping techniques: exercises



Try solving the following:
1. Print only even numbers from a list
2. Count occurrences of letters in a string using a loop
3. Print all keys and values from a dictionary
4. Merge two lists using zip()

````python
# Exercise 1
nums = [1,2,3,4,5,6]
for n in nums:
    if n % 2 == 0:
        print(n)
````

````python
# Exercise 2
text = "banana"
count = {}
for ch in text:
    count[ch] = count.get(ch, 0) + 1
print(count)
````

````python
# Exercise 3
student = {"name": "Alice", "age": 20}
for k, v in student.items():
    print(k, v)
````

````python
# Exercise 4
a = [1,2,3]
b = ["a","b","c"]

for x, y in zip(a, b):
    print(x, y)
````

### 2.6 List comprehension: Exercises

Create a list of squares of numbers from 1 to 20.

````python
# Solution

squares = [x**2 for x in range(1, 21)]

print(squares)
````

Generate a list of even numbers from 1 to 50.

````python
# Solution

even_numbers = [x for x in range(1, 51) if x % 2 == 0]

print(even_numbers)
````

From 1 to 50, create a list of numbers divisible by both 3 and 5.


````python
# Solution

numbers = [x for x in range(1, 51) if x % 3 == 0 and x % 5 == 0]

print(numbers)
````

From this list:

`nums = [10, 15, 20, 25, 30, 35]`

keep only numbers greater than 20.

````python
# Solution

nums = [10, 15, 20, 25, 30, 35]

greater_than_20 = [x for x in nums if x > 20]

print(greater_than_20)
````

Extract only vowels from the string:

`text = "list comprehensions make python concise"`

````python
# Solution
text = "list comprehensions make python concise"

vowels = [char for char in text if char in "aeiou"]

print(vowels)
````
