##### Exercise 4

Given:

```python
words = ["fig", "elderberry", "kiwi", "watermelon", "date"]
```

Use `filter()` with a lambda to keep only the words longer than 4 letters.

````python
words = ["fig", "elderberry", "kiwi", "watermelon", "date"]

# Write your exercise here
def long_words(words):
    return list(filter(lambda word: len(word) > 4, words))

long_words(words)
````

````python
# Write your solution here

longer_words= []

for word in words:
  if len(word) > 4:
    longer_words.append(word)
  print(longer_words)
````
